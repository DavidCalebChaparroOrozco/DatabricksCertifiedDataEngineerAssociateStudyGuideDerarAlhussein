# Creating a databricks table

```python
%sql
CREATE TABLE workspace.default.users (
    id INT,
    name STRING,
    email STRING,
    age INT,
    created_at TIMESTAMP
) USING DELTA; -- How will it be saved? Here we tell it that it will be saved in deltalake format
```

# Insert Data

```python
%sql
INSERT INTO workspace.default.users VALUES(
    1,
    'Caleb',
    'caleb@gmail.com',
    28,
    current_timestamp() -- Function to get the current timestamp
)
```

```python
%sql
INSERT INTO workspace.default.users VALUES
    (2, 'Bender', 'bender@gmail.com', 30, current_timestamp()),
    (3, 'Jane', 'jane@gmail.com', 25, current_timestamp()),
    (4, 'Mike', 'mike@gmail.com', 35, current_timestamp());
```

## How do we look at it?

```python
%sql
SELECT *
FROM workspace.default.users
```

## Update Data

```python
%sql
UPDATE workspace.default.users
SET email = "caleb@outlook.com"
where id = 1
```

# DESCRIBE DETAIL users

```python
%sql
DESCRIBE DETAIL users;
```

```python
%sql
SELECT * 
FROM workspace.default.users VERSION AS OF 1;
```

```python
%sql
SELECT * 
FROM workspace.default.users TIMESTAMP AS OF '2026-02-15 22:51:00';
-- WHERE created_at BETWEEN '2026-02-15 22:16:00'
--                       AND '2026-02-15 23:05:00'
```

In a very extreme case, a trainee performed a DELETE operation without the WHERE clause. How do we solve this? Here's a step-by-step guide:

```python
%sql
DELETE FROM 
workspace.default.users;
```

```python
%sql
SELECT *
FROM workspace.default.users;
```

```python
%sql
DESCRIBE HISTORY users;
```

```python
%sql
RESTORE TABLE workspace.default.users TO VERSION AS OF 2;
```

```python
%sql
SELECT *
FROM workspace.default.users;
```

```python
%sql
DESCRIBE HISTORY users;
```
